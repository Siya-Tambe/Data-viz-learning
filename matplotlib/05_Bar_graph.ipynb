```python
import matplotlib.pyplot as plt
```

```python
# Create a data for plotting
values = [5,6,3,7,2]
names = ["A","B","C","D","E"]
```

```python
plt.bar(names, values, color = 'orange')

'''Here first tuple shows lable of x-axis and second tuple shows label of y-axis'''
plt.show()
```
