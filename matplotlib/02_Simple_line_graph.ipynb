```python
# importing required module
import matplotlib.pyplot as plt

# x axis values
x = [1,2,3]

# corresponding y axis values
y = [2,4,1]
```

```python
# plotting the points
plt.plot(x,y);

# naming the x-axis
plt.xlabel("X-axis")

# naming the y-axis
plt.xlabel("Y-axis")

# giving a title to my graph
plt.title("My first graph")

# function to show the plot
plt.show()
```
